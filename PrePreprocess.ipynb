```python
import pandas as pd

df = pd.read_csv("books_with_categories.csv")

print(df.columns)
print(df.shape)

# Keep the required columns
df = df.rename(columns={
    "title": "Book_Name",
    "description": "Description",
    "categories": "Genre"
})

df["Tags"] = df["Genre"]

df = df[[
    "Book_Name",
    "Description",
    "Genre",
    "Tags"
]]

df = df.dropna(subset=["Book_Name", "Description"])

df.to_csv("Book_Recommendation_Dataset.csv", index=False)

print(df.head())
print(df.shape)
```
